# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# 🛠️ TOOL 3 (Bonus): Word Counter
# A simple extra tool so the agent can do more than 2 things.

def word_counter(text: str) -> int:
    """Count the number of words in a piece of text."""
    try:
        return len(text.split())
    except Exception:
        return 0
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use Calculator Tool
- If query contains "keywords" → use Keyword Extractor Tool
- If query contains "count words" → use Word Counter Tool (bonus tool)
- Else → general response

We also add **basic logging** (bonus) so we can see which route the agent picked for every query, and **try/except** error handling so a bad query never crashes the whole program.


````python
# 🤖 AGENT FUNCTION (IMPLEMENTED)

import re

def agent(query: str):
    """
    Very simple single-agent router.
    Looks at keywords in the query to decide which tool to call.
    Always returns a dict in the format: {"type": ..., "result": ...}
    """
    query_lower = query.lower()

    try:
        # ---- Route 1: Calculator ----
        if "calculate" in query_lower:
            print(f"[LOG] Query -> '{query}' | Routed to: Calculator Tool")

            # remove the trigger word, keep only characters a math expression can have
            expression = query_lower.replace("calculate", "")
            expression = re.sub(r"[^0-9+\-*/(). ]", "", expression).strip()

            if expression == "":
                return {"type": "error", "result": "No valid math expression found."}

            calc_result = calculator(expression)
            if calc_result == "Error in calculation":
                return {"type": "error", "result": calc_result}

            return {"type": "calculation", "result": calc_result}

        # ---- Route 2: Keyword Extractor ----
        elif "keywords" in query_lower:
            print(f"[LOG] Query -> '{query}' | Routed to: Keyword Extractor Tool")

            # strip out the instruction words, keep the actual sentence
            text = query_lower
            for trigger_word in ["extract", "keywords", "from"]:
                text = text.replace(trigger_word, "")
            text = text.strip()

            if text == "":
                return {"type": "error", "result": "No text found to extract keywords from."}

            keywords = extract_keywords(text)
            return {"type": "keywords", "result": keywords}

        # ---- Route 3 (Bonus): Word Counter ----
        elif "count words" in query_lower:
            print(f"[LOG] Query -> '{query}' | Routed to: Word Counter Tool")

            text = query_lower.replace("count words", "").replace("in", "", 1).strip()
            if text == "":
                return {"type": "error", "result": "No text found to count words in."}

            count = word_counter(text)
            return {"type": "word_count", "result": count}

        # ---- Route 4: General Response ----
        else:
            print(f"[LOG] Query -> '{query}' | Routed to: General Response")
            return {
                "type": "general",
                "result": f"I received your query: '{query}'. How can I assist further?"
            }

    except Exception as e:
        # Basic error handling so the agent never crashes on a bad input
        return {"type": "error", "result": f"Something went wrong: {str(e)}"}
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?",
    "Count words in the quick brown fox jumps over the lazy dog",
    "Calculate"   # edge case: no expression given, should return an error
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    print("Response:", agent(user_input))
````
